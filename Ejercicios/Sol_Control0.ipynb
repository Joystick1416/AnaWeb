# Control 0 — N1: Números similares

Dos enteros positivos `a` y `b` (con `a < b`, misma cantidad de dígitos) son **similares** si `b` se puede obtener moviendo algunos dígitos del final de `a` hacia el frente, sin cambiar el orden entre ellos.

Ejemplos dados en el enunciado: `(12, 21)`, `(756, 675)`, `(12345, 34512)`.

Vamos a construir la solución paso a paso (ver también [Teoría/Fundamentos_Control0.md](../Teor%C3%ADa/Fundamentos_Control0.md)).

## Paso 1: representar los números como texto

Convertimos ambos números a `str` para poder trabajar con sus dígitos como si fueran una secuencia de caracteres.

```python
a = 12345
b = 34512

sa = str(a)
sb = str(b)

print(sa, sb)
```

## Paso 2: validar que tengan la misma cantidad de dígitos

El enunciado dice que ambos números tienen la misma cantidad de dígitos. Igual conviene verificarlo: si no coinciden en longitud, **no pueden** ser similares (no hay forma de rotar dígitos y cambiar cuántos hay).

```python
misma_longitud = len(sa) == len(sb)
print(misma_longitud)
```

## Paso 3: aplicar el truco de rotación

"Mover dígitos del final hacia el frente" es exactamente una **rotación** del string. El truco clásico: si concatenamos `sa` consigo mismo (`sa + sa`), ahí quedan escritas, una detrás de otra, **todas** las rotaciones posibles de `sa`.

Entonces: `sb` es una rotación de `sa` si y solo si `sb` aparece como substring de `sa + sa`.

```python
doble = sa + sa
print("sa + sa =", doble)

es_rotacion = sb in doble
print("¿sb está dentro de sa+sa? ->", es_rotacion)
```

## Paso 4: juntar todo en una función

Ahora empaquetamos los pasos 1-3 en una función reutilizable `son_similares(a, b)`.

```python
def son_similares(a, b):
    sa, sb = str(a), str(b)
    if len(sa) != len(sb):
        return False
    return sb in (sa + sa)
```

## Paso 5: probar con los ejemplos del enunciado

```python
casos = [(12, 21), (756, 675), (12345, 34512)]

for a, b in casos:
    print(f"({a}, {b}) -> {son_similares(a, b)}")
```

## Bonus: ¿y un caso que NO sea similar?

`123` tiene como rotaciones posibles: `123`, `231`, `312`. `321` no está entre ellas, así que debería dar `False`.

```python
print(son_similares(123, 321))  # esperado: False
```
